```python
create_sql = """
CREATE TABLE IF NOT EXISTS specimens (
    specimen_id            VARCHAR(10) PRIMARY KEY,
    collection_date        DATE,
    age_group              VARCHAR(10),
    sex                    VARCHAR(10),
    department             VARCHAR(20),
    specimen_type          VARCHAR(20),
    test_requested         VARCHAR(30),
    collection_shift       VARCHAR(15),
    collector_role         VARCHAR(30),
    rejection_status       VARCHAR(15),
    rejection_reason       VARCHAR(30),
    turnaround_time_min    FLOAT NULL,
    flag_dept_sex_mismatch BOOLEAN,
    flag_dept_age_mismatch BOOLEAN
);
"""

with engine.connect() as conn:
    conn.execute(text(create_sql))
```
